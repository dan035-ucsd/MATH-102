```python
import numpy as np
import matplotlib.pyplot as plt
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.set_xlabel(r'$t$')
ax.set_ylabel(r'$x$')
ax.set_zlabel(r'$u(x,t)$')
# ax.set_title(r"D'Alembert soln for $\phi(x)=e^{-x^2}$ and $\psi(x)=0$ with $c=1$")
# ax.set_title(r"D'Alembert soln for $\phi(x)=0$ and $\psi(x)=\cos x$ with $c=1$")
ax.set_title(r"D'Alembert soln for $\phi(x)= 0$ and $\psi(x)=e^{-|x|}$ with $c=1$")
ax.set_box_aspect(None, zoom=1)
ax.view_init(elev=30, azim=-190, roll=0)

c = 0.5

X = np.linspace(-5, 5, 100)
Y = np.linspace(0, 10, 100)
Z = np.zeros((len(X), len(Y)))
# Z = 1/2 * (np.exp( - (X - c * Y)**2) + np.exp( - (X + c * Y)**2))
# Z = 1/(2 * c) * (np.sin(X + c * Y) - np.sin(X - c * Y))
for i in range(len(X)):
    for j in range(len(Y)):
        if X[i] < -c * Y[j]:
            Z[i, j] = 1/(2 * c) * (np.exp(X[i] + c * Y[j]) - np.exp(X[i] - c * Y[j])) 
        elif X[i] < c * Y[j] and X[i] > -c * Y[j]:
            Z[i, j] = 1/(2 * c) * (- np.exp(-(X[i] + c * Y[j])) - np.exp(X[i] - c * Y[j])) + 1/c
        elif X[i] > c * Y[j]:
            Z[i, j] = 1/(2 * c) * (-np.exp(-(X[i] + c * Y[j])) + np.exp(-(X[i] - c * Y[j])))

X, Y = np.meshgrid(X, Y)
plt.xticks([])
plt.yticks([])
ax.plot_surface(X, Y, Z, cmap='viridis')
plt.show()
```
