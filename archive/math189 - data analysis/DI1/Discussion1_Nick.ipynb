Part (f) asks us to create a new qualitative variable called `Elite`, which denotes whether more than 50% of a school's incoming class is from the top 10% of their high school. The given code to do this is...

```python
college['Elite'] = pd.cut(college['Top10perc'], [0,0.5,1], labels=['No', 'Yes'])
```

We then want to use `value_counts` to see how many elite universities there are:

```python
college["Elite"].value_counts()
```

This seems wrong... what happened?

```python
college["Top10perc"]
```

It seems like `Top10perc` contains _percentages_ rather than _proportions_, so we can fix this by doing...

```python
college['Elite'] = pd.cut(college['Top10perc']/100, [0,0.5,1], labels=['No', 'Yes'])
```

```python
college["Elite"].value_counts()
```

Finally, we use `boxplot()` to view side-by-side boxplots of `Outstate` versus `Elite`

```python
college.boxplot("Outstate","Elite")
```

For part (g), we are asked to use `plot.hist()` to produce histograms with differing numbers of bins for a few of the quantitative variables. For just one histogram, we can update pass the number of bins as an argument:

```python
college["Outstate"].plot.hist(bins=20)
```

For multiple histograms, we can filter multiple columns:

```python
college[["Apps", "Accept", "Enroll", "Outstate"]].hist(bins=20,layout=(2, 2), figsize=(10,8))
```

Or we can use `plt.subplots` to achieve a similar effect:

```python
import matplotlib.pyplot as plt
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].hist(college["Apps"], bins=20)
axes[0, 0].set_title("Apps")

axes[0, 1].hist(college["Accept"], bins=20)
axes[0, 1].set_title("Accept")

axes[1, 0].hist(college["Enroll"], bins=20)
axes[1, 0].set_title("Enroll")

axes[1, 1].hist(college["Outstate"], bins=20)
axes[1, 1].set_title("Outstate")

plt.tight_layout()
plt.show()
```
