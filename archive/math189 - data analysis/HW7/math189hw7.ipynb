```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

X1 = [1,1,0,5,6,4]
X2 = [4,3,4,1,2,0]
Y = rng.choice(2, 6)
Colors = {0: 'blue', 1: 'red'}

def clustering(criterion):
    sumX1 = 0
    sumX2 = 0
    sumY = 0
    for i in range(len(Y)):
        if Y[i] == criterion:
            sumX1 += X1[i]
            sumX2 += X2[i]
            sumY += 1
    X1cluster = [sumX1 / sumY]
    X2cluster = [sumX2 / sumY]
    return X1cluster, X2cluster

fig, ax = plt.subplots(2,1, figsize=(4, 6))
fig.suptitle('K-Means Clustering')
fig.supxlabel('X1')
fig.supylabel('X2')
fig.tight_layout()

for i, j  in enumerate(Y):
    ax[0].scatter(X1[i], X2[i], color=Colors[j])

X1cluster0, X2cluster0 = clustering(0)
X1cluster1, X2cluster1 = clustering(1)
ax[0].set_title('Step 1')
ax[0].scatter(X1cluster0, X2cluster0, color='blue', marker='X', s=100, label='Cluster 0')
ax[0].scatter(X1cluster1, X2cluster1, color='red', marker='X', s=100, label='Cluster 1')
print(Y)

for i in range(len(Y)):
    dist0 = np.sqrt((X1[i] - X1cluster0[0])**2 + (X2[i] - X2cluster0[0])**2)
    dist1 = np.sqrt((X1[i] - X1cluster1[0])**2 + (X2[i] - X2cluster1[0])**2)
    if dist0 < dist1:
        Y[i] = 0
    else:
        Y[i] = 1

for i, j  in enumerate(Y):
    ax[1].scatter(X1[i], X2[i], color=Colors[j])

X1cluster0, X2cluster0 = clustering(0)
X1cluster1, X2cluster1 = clustering(1)
ax[1].set_title('Step 2')
ax[1].scatter(X1cluster0, X2cluster0, color='blue', marker='X', s=100, label='Cluster 0')
ax[1].scatter(X1cluster1, X2cluster1, color='red', marker='X', s=100, label='Cluster 1')
```
