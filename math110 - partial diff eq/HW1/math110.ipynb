```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
x = np.linspace(-5, 5, 100)
y1 = -1/2 * x
y2 = -1/2 * x + 1
y3 = -1/2 * x - 1
y4 = -1/2 * x + 2
y5 = -1/2 * x - 2
ax.plot(x, y1)
ax.plot(x, y2)
ax.plot(x, y3)
ax.plot(x, y4)
ax.plot(x, y5)
ax.legend()
ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y = -1/2 * x + c')
```

```python
fig, ax = plt.subplots()
x = np.linspace(-5, 5, 100)
y1 = np.arctan(x)
y2 = np.arctan(x) + 1
y3 = np.arctan(x) - 1
y4 = np.arctan(x) + 2
y5 = np.arctan(x) - 2
ax.plot(x, y1)
ax.plot(x, y2)
ax.plot(x, y3)
ax.plot(x, y4)
ax.plot(x, y5)
ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y = arctan(x) + c')
```
