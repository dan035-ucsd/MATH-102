```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
x = np.linspace(-5, 5, 100)
y1 = -1/2 * x
y2 = -1/2 * x + 1
y3 = -1/2 * x - 1
y4 = -1/2 * x + 2
y5 = -1/2 * x - 2
ax.plot(x, y1)
ax.plot(x, y2)
ax.plot(x, y3)
ax.plot(x, y4)
ax.plot(x, y5)
ax.legend()
ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y = -1/2 * x + c')
```

```python
fig, ax = plt.subplots()
x = np.linspace(-5, 5, 100)
y1 = np.arctan(x)
y2 = np.arctan(x) + 1
y3 = np.arctan(x) - 1
y4 = np.arctan(x) + 2
y5 = np.arctan(x) - 2
ax.plot(x, y1)
ax.plot(x, y2)
ax.plot(x, y3)
ax.plot(x, y4)
ax.plot(x, y5)
ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y = arctan(x) + c')
```

```python
fig, ax = plt.subplots()
x = np.linspace(-5, 5, 100)
y1 = np.sqrt(x**2)
y2 = np.sqrt(x**2) + 1
y3 = np.sqrt(x**2) - 1
y4 = np.sqrt(x**2) + 2
y5 = np.sqrt(x**2) - 2
y6 = -np.sqrt(x**2)
y7 = -np.sqrt(x**2) + 1
y8 = -np.sqrt(x**2) - 1
y9 = -np.sqrt(x**2) + 2
y10 = -np.sqrt(x**2) - 2

ax.plot(x, y1, color='blue')
ax.plot(x, y2, color='orange')
ax.plot(x, y3, color='green')
ax.plot(x, y4, color='red')
ax.plot(x, y5, color='purple')

ax.plot(x, y6, color='blue')
ax.plot(x, y7, color='orange')
ax.plot(x, y8, color='green')
ax.plot(x, y9, color='red')
ax.plot(x, y10, color='purple')

ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y = arctan(x) + c')
```
