```python
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt

# Creating dataset
x = np.outer(np.linspace(-3, 3, 32), np.ones(32))
y = x.copy().T # transpose
z = np.sin(x) + np.cos(y)

# Creating figure
fig = plt.figure(figsize =(14, 9))
ax = plt.axes(projection ='3d')

# Creating plot
ax.plot_surface(x, y, z)

# show plot
plt.show()
```

```python
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt


# Creating dataset
x = np.outer(np.linspace(-3, 3, 32), np.ones(32))
y = x.copy().T # transpose
z = (np.exp((-x **2 - y **2)/2))

# Creating figure
fig = plt.figure(figsize =(14, 9))
ax = plt.axes(projection ='3d')

# Creating plot
ax.plot_surface(x, y, z)

# show plot
plt.show()
```

```python
x = np.linspace(-5, 5, 100)
t = 100 # time
D = 1 # diffusion coefficient
r = 1 # reaction rate
u0 = 1 # initial concentration

y = u0 / (2 * np.sqrt(np.pi * D * t)) * np.exp(r * t - (x ** 2) / (4 * D * t))

# Creating figure
ax = plt.axes()
# Creating plot
ax.plot(x, y)
ax.set_title('Reaction-Diffusion Solution at t=' + str(t))
ax.set_xlabel('x')
ax.set_ylabel('u(x, ' + str(t) + ')')
ax.legend(['D=' + str(D) + ', r=' + str(r) + ', u0=' + str(u0)])
```

```python
L = 1 # birth rate
t = np.linspace(0, 5, 100)
y1 = np.exp(-L * t) * (1 - np.exp(-L * t)) ** (1 - 1)
y2 = np.exp(-L * t) * (1 - np.exp(-L * t)) ** (2 - 1)
y3 = np.exp(-L * t) * (1 - np.exp(-L * t)) ** (3 - 1)
# Creating figure
ax = plt.axes()
# Creating plot
ax.plot(t, y1, label=r'$\varphi_1(t)$')
ax.plot(t, y2, label=r'$\varphi_2(t)$')
ax.plot(t, y3, label=r'$\varphi_3(t)$')
ax.set_title(r'Yule Process Probability w/ $L=1$')
ax.set_xlabel(r'$t$')
ax.set_ylabel(r'$\varphi_k(t)$')
ax.legend()
plt.show()
```
