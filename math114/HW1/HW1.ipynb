```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# add plot of unit circle
fig, ax = plt.subplots(figsize=(6, 6))
circle = patches.Circle((0, 0), 1, edgecolor='blue', facecolor='none', linewidth=2)
ax.add_patch(circle)
ax.set_xlim(-1, 1)
ax.set_ylim(-1, 1)

# generate random points and plot them
rng = np.random.default_rng(1221)
N = 100000
x = rng.uniform(-1, 1, N)
y = rng.uniform(-1, 1, N)
inside_circle = x**2 + y**2 <= 1
ax.scatter(x[inside_circle], y[inside_circle], color='green', s=1)
ax.scatter(x[~inside_circle], y[~inside_circle], color='red', s=1)

# estimate pi
tau = inside_circle.sum()
pi_est = 4 * tau / N

# error term
error = abs(np.pi - pi_est)

(pi_est, error)
```
