```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# add plot of unit circle
fig, ax = plt.subplots(figsize=(6, 6))
circle = patches.Circle((0, 0), 1, edgecolor='green', facecolor='none', linewidth=2)
ax.add_patch(circle)
ax.set_xlim(-1, 1)
ax.set_ylim(-1, 1)

# generate random points and plot them
rng = np.random.default_rng(1221)
N = 10000
x = rng.uniform(-1, 1, N)
y = rng.uniform(-1, 1, N)
inside_circle = x**2 + y**2 <= 1
ax.scatter(x[inside_circle], y[inside_circle], color='green', s=1)
ax.scatter(x[~inside_circle], y[~inside_circle], color='purple', s=1)

# estimate pi
tau = inside_circle.sum()
pi_est = 4 * tau / N

# error term
error = abs(np.pi - pi_est)

(pi_est, error)
```

```python
rng = np.random.default_rng(218)
N = 1000
u = rng.uniform(0, 1, N)
x = []
for entry in u:
    if entry < 0.5:
        x.append(np.sqrt(2*entry))
    else:
        x.append(2 - np.sqrt(2 - 2*entry))

def f(x):
    if 0 <= x <= 1:
        return x
    elif 1 < x <= 2:  
        return 2 - x

x_series = np.arange(0, 2, 0.01)
y_series = np.vectorize(f)(x_series)

plt.hist(x, bins=100, density=True, edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Target PDF')
plt.plot(x_series, y_series)

plt.show()
```

```python
x = rng.uniform(0, 2, N)
y = rng.uniform(0, 0.5, N)
x_accept = []
for i in x:
    for j in y:
        if j < f(i):
            x_accept.append(i)
plt.hist(x_accept, bins=100, density=True, edgecolor='black')
```
