```python
import numpy as np

N = 100

def crude_monte_carlo_integration():
    X_n = np.random.uniform(0, 4, N)
    sum = 0
    for x in X_n:
        sum += 4 * (np.sqrt(x) * np.exp(-x))
    I = 1 / N * sum
    return I

def importance_sampling_integration():
    U = np.random.uniform(0, 1, N)
    Y_n = -np.log(1 - U * (1 - np.exp(-4)))

    sum = 0
    for y in Y_n:
        sum += (np.sqrt(y) * np.exp(-y)) / (np.exp(-y) / (1 - np.exp(-4)))
    I = 1 / N * sum
    return I

def samp_var(X):
    sum = 0
    for i in range(len(X)):
        sum += (X[i] - np.mean(X)) ** 2
    return sum / (len(X) - 1)

results_mc = []
for i in range(1, 101):
    results_mc.append(crude_monte_carlo_integration())
var_mc = samp_var(results_mc)
crude_monte_carlo_integration(), var_mc

results_is = []
for i in range(1, 101):
    results_is.append(importance_sampling_integration())
var_is = samp_var(results_is)
importance_sampling_integration(), var_is
```
